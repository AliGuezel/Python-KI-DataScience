## Copies and Views
Ein Numpy-Array kann (in einem "view") direkt über den Datenbuffer geändert werden ohne das Daten kopiert werden müssen oder es kann eine Kopie ("copy") erstellt werden, so daß das Original erhalten bleibt.

* Ein "View" entspricht einem "Teil" des Original Arrays. Änderungen am View oder Original werden auf beiden ausgeführt, da sie den selben Speicherbereich referenzieren
    * Views werden bei Indexierung automatisch erzeugt (keine Kopie !)
    * mit view() lässt sich explizit ein View erzeugen
* Eine "Copy" ist eben eine Kopie des Speicherbereichs an eine anderen Speicher-Position. Änderungen beeinflussen sich daher nicht gegenseitig.
    * eine Kopie kann über .copy() erzeugt werden
    * oder mittels Array-Konstruktor: `np.array(x, copy=True)`


```python
#Beispiel View:
import numpy as np
x = np.arange(10)
print(x)
y = x[1:3]  # creates a view
print(y)
x[1:3] = [10, 11]
print(x)
print(y) # Änderungen auf dem Original spiegeln sich auch im View wieder

y[0:2] = [22,33]
print(x) # Ändereungen am View spiegeln sich im Original wieder

z = x.view() # erzeugt einen views
z[0:1] = [100] # klammern kann man weglassen
print(x)

```

```python
#Beispiel Copy:
import numpy as np
x = np.arange(10)
print(x)

z = x.copy() # erzeugt eine Kopie
z[0:1] = [100] #
print(x)

z = np.array(x, copy=True) # erzeugt eine Kopie
z[0:1] = [1001] #
print(z)
print(x)
```

```python
import numpy as np

a = [[1, 2], [3, 4]]
print(a)

x = np.array(a)
y = x
z = np.array(x, copy=True) 

z = np.copy(x)

x[0,0] = 10 # 2d array !
y[1,1] = 4


print("-----")
print(x,y,z,sep="\n------\n")
```
