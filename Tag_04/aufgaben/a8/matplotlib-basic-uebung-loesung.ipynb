3.) Zeichne zusätzlich zu "2.b)" den __Simple-Moving-Average__ (SMA) von 100 Tagen ein.   
Also den __gleitenden Durchschnitt__ der letzten 100 Tage der wie folgt berechnet wird:  
df['Close'].rolling(100).mean()  

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('data/MSFT.csv',parse_dates=True,index_col=0)

df['Close'].plot()

df['Close'].rolling(100).mean().plot()
#vgl.: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.rolling.html

plt.show()
```
