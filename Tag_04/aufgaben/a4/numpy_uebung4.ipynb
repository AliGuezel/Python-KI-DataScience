

#### Ein 3x3 LGS mit Numpy lösen

4x + 3y -1z = 2

-3x -4y +5z = -5

-2x +2y +z = 6

Numpy biete die Funktion `linalg.solve(a, b)` um lineare Gleichungssysteme zu lösen.
vgl.:  
<https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html>

Also sind die Voraussetzungen um die Funktion anzuwenden:
1. die Funktion kann nur verwendet werden, wenn das LGS ***eine*** Exakte Lösung hat.  
(aber ein LGS kann __keine__,__eine__ oder __unendlich__  viele Lösungen haben)
2. die matrix einen vollen Rang hat  
(https://studyflix.de/mathematik/rang-einer-matrix-1785)
3. das LGS der Matrix-Gleichung ax = b entspricht

1\. und 2\. sind erfüllt. 

zu 3.:
Eine Matrix-Gleichung der Form Ax = b für das obige LGS ist: 

$\left[ \begin{array}{rrr}
4 & 3 & -1 \\ 
-3 & -4 & 5 \\ 
-2 & 2 & 1 \\ 
\end{array}\right]$
$\cdot$
$\left( \begin{array}{rrr}
x \\ 
y \\ 
z \\ 
\end{array}\right)$
= $\left( \begin{array}{rrr}

-5 \\
6 \\
\end{array}\right)$

<=>
