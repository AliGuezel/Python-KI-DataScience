# Matplotlib
 
* matplotlib ist eine Python Bibliothek zur Erstellung von Graphen 
* matplotlib lässt sich in Python Skripten aufgerufen, aber auch innerhalb der Jupyter Notebooks oder auch auf Web-Applicationen
* Es lassen sich vielseite Diagramme erstellen wie: __Plot__, __Histogramm__, Spektrum, __Balkendiagramm__, Kuchendiagramm, Scatter u.a.
* Die Matplot-API: <https://matplotlib.org/3.1.1/api/>

# Ein Beispiel

````python
import numpy as np
import matplotlib.pyplot as plt

# definition der (x,y) Wertpaare
x = np.arange(0,20, 0.01) # 
print(x)
y = np.sin(x)
#print(type(x))

# schreiben der (x,y)-Wertpaare 
plt.plot(x, y)
plt.show()


````

# ein erstes Beispiel für "Plot" (geeignet z.B für Mathematische Funktionen)

* also f(x) = y 
* wir benötigen also x-wWrte für die wir den  y-Wert mittels der Funktions-Gleichung ausrechnen.    
- also wenn f(x) = sin(x) gilt:    
  f(1)= sin(1) daraus folgt daß der Graph an der Stelle 1 den y Wert sin(1) hat  und sin(1)= 0.8414709848078965  
  f(2)= sin(2) = 0.9092974268256817  
  ..

* x und y sind numpy-arrays mit dtype = float64
* man könnte auch linspace statt arrange verwenden, oder selber ein Numpy-Array manuell wert zuweisen

* Es können Achsen-Bezeichnungen so wie ein Titel für das Diagram angegeben
* die Position des Title kann über das "loc"-Argument gesetzt werden: loc={'center', 'left', 'right'}
* auch Fonts können definiert werden über das "fontdict"-Argument 


````python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 50) # was ist der Unterschied zwischen den beiden Funktionen np.arange & linspace ?

y = np.add(np.power(x,2),0)

#z = np.sin(x)
#q = np.tan(x)
#plt.plot(x, y)
#plt.plot(x, z)
plt.plot(x,y)


plt.title("x hoch 2", loc="left")
plt.xlabel("x-Achse")

font = {'family':'serif','color':'green','size':30}
plt.ylabel("y-Achse",fontdict= font)

plt.show() # 
````

* es können mehrere Daten-Sets auf einem "Plot" platziert werden
* um das fertige Plot anzuzeigen muss am Ende die Funktion show() aufgerufen werden. (!?)
* es können andere Marker gewählt mittels den Funktionsparamenter "marker"
* der angezeigte y-Achsen wird von Matplotlib automatisch gewählt

````python
import numpy as np
import matplotlib.pyplot as plt
#x = np.arange(0, 3 * np.pi, 0.2) #
x = np.linspace(0., 10, 100) # was ist der Unterschied zwischen den beiden Funktionen np.arange & linspace ?
y = np.sin(x)
z = np.cos(x)

#
#plt.plot(x, y, marker="*")
#plt.plot(x, y, marker="D")
plt.plot(x, y, "r+")
plt.plot(x,z,':')
#plt.plot(x, y, marker="r+") #  != ?


plt.show()
````

* Eine Liste der möglichen Marker findet man unter: https://matplotlib.org/stable/api/markers_api.html
* Ein paar Beispiele: "D" = "Diamond", "v" = "Triangle Down", "*" = Star
* als drittes Argument kann ein String übergeben werden der die Funktion-Darstellung der Linie beeinflusst: plt.plot(x, y, "r+")


* Achtung: plt.plot(x, y, "r+") ist != plt.plot(x, y, marker="+"), denn das Argument "marker" bezieht sich nur auf den Marker und ist nicht das gleiche wie die "Format-Strings" als drittes Argument. Der Signature-String ist eine Kurzschreibweise für eine Kombination aus verschiedenen Named-Arguments
* Es gibt noch weitere Named-Arguments für die Darstellung des Plot z.B.:
plot(x, y, color='green', marker='o', linestyle='dashed', linewidth=2, markersize=12)
entspricht:
plot(x, y, 'go--', linewidth=2, markersize=12)

## Beispiel Histogramm
* Histogramme werden für Häufigkeitsverteilungen verwendet
* 

````python
import numpy as np
import matplotlib.pyplot as plt

mittelwert = 100
standard_abweichung = 30
size = 10000
 
### 1. Normalverteilung
values = np.random.normal(mittelwert, standard_abweichung, size)
# vgl. https://numpy.org/doc/stable/reference/random/generated/numpy.random.normal.html 

### 2. Gleichverteilung:
#values = np.random.random(1000) 

#print(values) 

# histogram
plt.hist(values, 100) #  100 "bins" <=> Behälter

 
plt.show()

````

## Beispiel Histogram mit Pandas 

* seaborn wird verwendet um attraktive Plots zu generieren
* https://seaborn.pydata.org


````python
import pandas as pd

data = pd.read_csv('csv/president_heights.csv')
#print(data.head(4)) # index kann als "order"-Spalte definiert werden siehe Pandas

heights = np.array(data['height(cm)']) 

import seaborn; seaborn.set()  # plot style

plt.hist(heights,20)
plt.title('Height Distribution of US Presidents')
plt.xlabel('height (cm)')
plt.ylabel('number')


````

## Beispiele für Bar Charts

````python
import matplotlib.pyplot as plt

ywerte = [4, 7, 1, 9, 5, 2, 8]
xwerte = ["SPD", "CDU/CSU", "AFD", "Grüne", "Linke", "Freie Wähler", "Basis"]

plt.bar(xwerte, ywerte)
plt.xlabel("X-Werte")
plt.ylabel("Y-Werte")
plt.show()
````
