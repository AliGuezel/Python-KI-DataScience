# Markdown

* Markdown ist eine vereinfachte Auszeichnungssprache (Markup-Language) mit dem Ziel einer leicht lesbaren Ausgangsform bereits vor der Konvertierung.  
* Die Markdown-Konvertierungssoftware wandelt Markdown-Text in gültiges und W3C-konformes XHTML um.

<hr>

* Markup Language bezeichnet eine maschinenlesbare Auszeichnungssprache, zur Formatierung und Gliederung von Texten und anderen Daten.

### Eine kleine Einführung in Markdown


## Überschriften können mit "\#" vor der Zeile definiert werden

Beispiele:

# \# Überschrift 1 
## \#\# Überschrift 2 
### \#\#\# Überschrift 3 
#### \#\#\#\# Überschrift 4 

\## Headline

## Auflistungen können mit  \* (Bullet) oder 1., 2., .. definiert werden. Einrückung wird mit 4 Leeren Zeichen definiert.

Bsp.:

* Listen-Element 1
* Listen-Element 2
    * Listen-Element 3

oder  

1. Listen-Element 1  
2. Listen-Element 2  
    3. Listen-Element 3



## Steuerzeichen wie die Raute \# können mit einen Backslash \ aufgehoben werden. Das gilt auch für den Backslash selber
* \\\*
* \\\'
* \\\_
* \\

*

Beispiel ***Kursiv*** => \*** Kursiv \***


## Fett und Kursiv können aus einer Kombination von \* oder \_ definiert werden:

* _kursiv_, __fett__, ___fettkursiv___ (\_kursiv\_, \_\_fett\_\_, \_\_\_fettkursiv\_\_\_)
* *kursiv*, **fett**, ***fettkursiv*** (\*kursiv\*, \*\*fett\*\*, \*\*\*fettkursiv\*\*\*)

## Bilder einfügen

Es gibt grundsätzlich zwei Möglichkeiten Bilder einzufügen:
* Inline:
        `![Alt-Text](/Pfad/zum/Bild.jpg)`  
        => `<img src="/Pfad/zum/Bild.jpg" alt="Alt-Text" title="" />`
        oder:
        `![Alt-Text](/Pfad/zum/Bild.jpg "optionaler Titel")`  
        => `<img src="/Pfad/zum/Bild.jpg" alt="Alt-Text" title="optionaler Titel" />`

* Referenz:
    * Definition: ``[ref1]: Pfad/zum/Bild.jpg  "Optionaler Titel"``
    * Ausgabe: ``![Alt-Text][ref1]``


#### Bsp.: Inline

![Tiger-Bild](img/tiger.jpg)

#### Bsp.: Referenz

[ref1]: img/icebear.png  "Eisbär png"
[ref2]: schmetterling.jpg  "Eisbär png"

![Eisbär][ref1]

* jede Tabellenzeile in einer eigenen Zeile.
* Tabellenzellen werden mithilfe des Pipezeichens | getrennt
* Die ersten beiden Zeilen einer Tabelle legen die Spaltenüberschriften und die Ausrichtung der Elemente in der Tabelle fest.
* : links => Text wird links ausgerichtet
* : rechts => Text wird rechts ausgerichtet
* : links & rechts => Text wird mittig ausgerichtet
* die Anzahl der - ist nicht relevant
* Zeilenumbruch innerhalb von Zellen mit <br/>



| Heading 1 | Heading 2 | Heading 3 |  
|:----------|:---------:|----------:|  
| Cell A1   | Cell A2   | Cell A3 |  
| Cell B1   | Cell B2   | Cell B3<br/>second line of text |
