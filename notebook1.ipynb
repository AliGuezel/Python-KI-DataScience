```python
## Diagramme können erstellt werden, dazu später mehr

import numpy as np
import matplotlib.pyplot as plt# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 3 * np.pi, 0.1)
y = np.cos(x)
z = np.sin(x)
print("Plot the points using matplotlib:")
plt.plot(x, y)
plt.plot(x, z,'r+')
plt.show()
```
