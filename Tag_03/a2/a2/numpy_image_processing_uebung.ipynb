## Hinweis:
In der verwendeten Environment müssen die Packages Pillow, Numpy, Jupyter installiert sein

```python
from PIL import Image #  "Python Image Library" Pillow Library https://python-pillow.org/
from IPython.display import Image as ShowImage # wird benötigt um das Bild innerhalb eines Notebooks auszugeben (Wegen Namens konflikt wird hier "ShowImage" verwendet)
import numpy as np 

im = np.array(Image.open('img/tiger.jpg')) # ein Bild wird geladen und in ein Numpy-Array konvertiert

print(type(im)) # Da es sich um ein Numpy- Array handelt, können alle Methoden auf Numpy-Arrays angewendet werden
print("dtype",im.dtype)
print("shape",im.shape)

# Es handelt sich um ein 3-d-Array: (427, 640, 3)
# 427 ist die Höhe des Bildes (anders als sonst üblich bei Bildbearbeitung ist die erste Koordinate die vertikale Position, die zweite die horizontale Position)
# 640 ist die Breite des Bildes
# die dritte Dimension stellt ein Pixel im RGB-Format dar: 
# Rotanteil R, der Grünanteil G und der Blauanteil B die Farbe beschreiben: Farbe = (R, G, B), dabei sind jeweils Werte von 0 bis 255 erlaubt
# [255,255,255] = Weiß, [0,0,0] = schwarz, ..

# Mittels Array-Slicing kann man z.B ein weißes Rechteck in das Bild zeichen:
im[50:85,50:175] = (255,255,255)


# Array in ein Bild umwandeln
pil_img = Image.fromarray(im)

# Das Bild speichern
pil_img.save('img/out/tiger.jpg') 
# Das manipulierte Bild wird in einem unterordner gespeichert. 
# Bei endung .jpg wird der Array auch gleich komprimiert, - daher können einzelne Pixel im Ergebnis nicht so aussehen wie erwartet.
# bitmap funktioniet nicht mit display
# Es wird nur gepseichert um es innerhalb des Notebooks wieder ausgeben zu können:

ShowImage(filename='img/out/tiger.jpg') 
```
