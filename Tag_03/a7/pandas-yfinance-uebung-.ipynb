yfinance steht für Yahoo-Finance. Es ist eine API um Börsenkurse und Informationen über Kurse und Unternehmen abzufragen

* führe den Pyhton-Code aus

Die Variable stock_data ist ein Pandas-Dataframe welche mit den entspechenden Methoden weiterverarbeitet werden kann
  
* zeichne einen Graph aus dem Schlusskurs, also der Spalte "Close" mittels "plot()"
  
* gib den Tages-Datensatz aus mit dem höchsten "Volume"
  
* gib den Tages-Datensatz aus mit dem höchsten Handelsspanne (maximum von High - Low)  
Hinweis: Erzeuge eine neue Spalte für die Handelsspanne

```python
# pip install yfinance numpy matplot

import numpy as np
import yfinance as yf
import matplotlib

def getStock(symbol : str) -> yf.Ticker:
    if symbol:
        try:
            stock = yf.Ticker(symbol)
        except Exception as e: 
            print(e)

    return stock

def getStockDataFromSymbol(symbol: str,period = '3y',interval= '1d'):  
    stock = getStock(symbol)

    return stock.history(period, interval)


symbol = "BTC-USD" 


stock_data = getStockDataFromSymbol(symbol,interval= '1d')
print(stock_data)
print(type(stock_data))




```

```python
#1.
stock_data['Close'].plot()

```

```python
#2.
idmax = stock_data['Volume'].idxmax()
print("max-volume: ")
print(stock_data.loc[idmax])

```

```python
#3.
stock_data['Range'] = stock_data['High'] - stock_data['Low']
idmax_range = stock_data['Range'].idxmax()
print("max-range: ")
print(stock_data.loc[idmax_range])
```
