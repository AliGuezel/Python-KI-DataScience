#### 2. Teste ob ALLE Elemente ungleich 0 sind und gebe diese Information aus:

```python
import numpy as np 
x1 = np.array([1, 2, 3, 4])

print(np.all(x1 != 0))
# code hier
x2 = np.array([1, 2, 0, 4])
# code hier
print(np.all(x2 != 0))

```

#### 3. Teste ob EIN Elemente ungleich 0 ist und gebe diese Information aus:

```python
import numpy as np 
x1 = np.array([1, 2, 3, 4])

# code hier
print(np.any(x1 != 0))

x2 = np.array([1, 2, 0, 4])
x2 = np.array([0, 0, 0, 0])# => false

# code hier
print(np.any(x2 != 0))


```
