#### Schreibe ein Program welches eine Pandas-Series in ein Python-Liste konvertiert und zusammen mit dem "type" ausgibt

```python
import pandas as pd
ds = pd.Series([2, 4, 6, 8, 10])

mylist = ds.tolist()
print(type(mylist),mylist)


```

#### Schreibe ein Program welches zwei Dataseries [2, 4, 6, 8, 10] und [1, 3, 5, 7, 9] Elementweise addiert, subtrahiert und multipliziert 

```python
import pandas as pd
ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 9])
ds = ds1 + ds2
print("Addition:")
print(ds)
print("Subtraktion:")
ds = ds1 - ds2
print(ds)
print("Multiplikation:")
ds = ds1 * ds2
```

#### Korrigiert in der folgenden Data-Series den Wert für Spitzbergen zu -5 und füge den Wert für Warschau = 25 hinzu im nachhinein über indexing (nicht bei der Konstuktion der Series) und gib die Series aus

```python
import pandas as pd

temperatures = pd.Series({'Berlin':30,'Rom':35,'Spitzbergen':5,'Kairo':40})
print(temperatures)
# 
temperatures['Spitzbergen'] = -5
temperatures['Warschau'] = 25
print(temperatures)


```

#### Lösche den Wert für "Kairo" mit der .drop() funktion
vgl.: https://pandas.pydata.org/docs/reference/api/pandas.Series.drop.html

```python
temperatures = temperatures.drop(labels=['Kairo'])
print(temperatures)
```

#### Berechne für folgende Series den Arithmetischen Mittelwert und den Median mit den entspechenden Pandas-Methoden

```python
import pandas as pd

ds = pd.Series([2.5,6.7,10.2,-3.0, 33.5])

print(ds.mean())
print(ds.median())
```
