## Jupyter Notbooks

* Notebooks sind Dokumente welche von der Jupyter Notebook App produziert werden.  
(Auch IDE's wie VS-Code sind in der Lage Jupyter Notebooks zu interpretieren, evt. wird eine Extension benötigt.  
VS-Code : "Jupyter Extension for Visual Studio Code)
* Sie enthalten sowohl Computer-Code (z.B. Python) als auch Rich-Text Elemente (Links, Absätze, Kursive Darstellung, .. )
* Somit lassen sich einerseits einfach Beschreibungen einer Analyse und der Resultate darstellen, wie auch ausführbaren Code für eine Analyse in einem Dokument präsentieren. 
* Ein Python-Notebook-Dokument hat die Dateiendung `.ipynb`
* Ein Notebook besteht aus aufeinander folgenden Darstellungs- oder Code-Zellen
* Die Darstellungs-Zellen können Markdown, HTML, u.a. darstellen
* Die Code-Zellen können Programm-Code ausführen und sind dabei nicht auf Python beschränkt
* Notebook Dokumente können in verschiedene andere Formate wie HTML oder PDF konvertiert werden

Beispiele für Python-Code:

```python

mystr = "hello World"
mylist = [23,4,34,5,3,5]
```

```python
# Code ist über mehrere aufeinander folgende Zellen gültig
mystr # handelt es sich in der Letzten Zeile einer Zelle um eine Variable wird diese ausgegeben
mylist # nur die letzte zeile wird ausgegeben
print(mylist) 
print(mystr) # will man mehrere Zeilen ausgeben kann man mit print() arbeiten
```

```python
# Mit print() lassen sich mehrere Ausgaben machen
print("zeile 1")
print("zeile 2")
```

```python
## Diagramme können erstellt werden, dazu später mehr

import numpy as np
import matplotlib.pyplot as plt # https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

# Compute the x and y coordinates for points on a sine curve
x = np.arange(0, 3 * np.pi, 0.1)
y = np.cos(x)
z = np.sin(x)
print("Plot the points using matplotlib:")
plt.plot(x, y)
plt.plot(x, z,'r+')
plt.show()
```
